```python
from skimage import data
import matplotlib.pyplot as plt
rocket_image = data.rocket()
def show_image(image, title="Image", cmap_type='gray'):
    plt.imshow(image, cmap=cmap_type)
    plt.title(title)
    plt.axis('off')
    plt.show()
show_image(rocket_image, "Rocket")
```

```python
import inspect

# List all public callables in skimage.data
image_funcs = [
    name for name, obj in inspect.getmembers(data, inspect.isfunction)
    if not name.startswith("_")
]

print(image_funcs)
```

```python
chelsea = data.chelsea()
print(type(chelsea))
show_image(chelsea, "Chelsea")
```

```python
#chelsea gray scale
from skimage import color
grayscale = color.rgb2gray(chelsea)
show_image(grayscale, 'chelsea grayscaled')
```

```python
# obtaining individual color intensity in an image
red = chelsea[:,:,0]
green = chelsea[:,:,1]
blue = chelsea[:,:,2]
plt.imshow(red, cmap='gray')
plt.title('Red')
plt.axis('off')
plt.show()
```

```python
#plot green
plt.imshow(green, cmap="gray")
plt.title('Green')
plt.axis('off')
plt.show()
```

```python
#plot blue
plt.imshow(blue, cmap="gray")
plt.title('Blue')
plt.axis('off')
plt.show()
```
