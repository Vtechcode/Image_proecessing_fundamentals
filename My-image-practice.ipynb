```python
import matplotlib.pyplot as plt
my_image = plt.imread(r"C:\Users\hp\Desktop\Documents\images\linkedin.jpg")
```

```python
def show_image(image, title='Image', cmap_type='gray'):
    plt.imshow(image, cmap=cmap_type)
    plt.title(title)
    plt.axis('off')
    plt.show()

show_image(my_image, "me")
```

```python
from skimage import color
grayscale = color.rgb2gray(my_image)
show_image(grayscale, "me-gray")
```

```python
#obtaining individual color intensity in an image
red = my_image[:,:,0]
green = my_image[:,:,1]
blue = my_image[:,:,2]
plt.imshow(red, cmap="gray")
plt.title('Red')
plt.axis('off')
plt.show()
```

```python
plt.imshow(green, cmap="gray")
plt.title('Green')
plt.axis('off')
plt.show()
```

```python
plt.imshow(blue, cmap="gray")
plt.title('Blue')
plt.axis('off')
plt.show()
```

```python
#flipping images
import numpy as np
#vertically
vertically_flipped = np.flipud(my_image)
show_image(vertically_flipped, 'vertically flipped image')
```

```python
# Threshold
#obtain the optimal threshold value
thresh = 127
#Apply thresholding to the image
binary = grayscale > thresh
#show the original and threshold
show_image(my_image, 'Original')
show_image(binary, 'thresholded')
```

```python
#Global uniform background
#import the otsu threshold function
from skimage.filters import threshold_otsu
#obtain the optimal threshold value
thresh = threshold_otsu(grayscale)
#Apply thresholding to the image
binary_global = grayscale > thresh
#show the original and binarized image
show_image(my_image, 'Original')
show_image(binary_global, 'Global threshoding')
```

```python
#local uneven background
#import the local threshold function
from skimage.filters import threshold_local
#set the block size to 35
block_size = 35
#obtain the optimal local thresholding
local_thresh = threshold_local(grayscale, block_size, offset=10)
#Apply local thresholding and obtain binary image
binary_local = grayscale > local_thresh
#show the original and binarized image
show_image(grayscale, 'Original')
show_image(binary_local, 'Local threshoding')
```
