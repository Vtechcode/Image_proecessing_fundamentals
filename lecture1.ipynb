```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as img

plt.title("Apollo Landing Site")
image = img.imread("images/Apollo-15-landing-site.jpg")
plt.imshow(image, cmap='gray')
plt.show()
```

```python
# maximum intensity
np.max(image)
```

```python
# minimum intensity
np.min(image)
```

```python
# size
image.shape
```

```python
# Histogram of pixel values
plt.hist(image.flatten(), 256)
plt.title('Histogram of pixel intensity values')
plt.xlabel('Intensity')
plt.ylabel('Count')
```
