## Adding noise to an image in scikit-image

```python
# load image
from skimage import data
from matplotlib import pyplot as plt
rocket_image = data.rocket()

# visualizing images
def show_image(image, title='Image', cmap_type='gray'):
    plt.imshow(image)
    plt.title(title)
    plt.axis('off')
    plt.show()

show_image(rocket_image, "Rocket")
```

```python
from skimage.util import random_noise
noisy_rocket = random_noise(rocket_image)
show_image(noisy_rocket, "Rocket with noise")
```

# Denoising using Total Variation

```python
from skimage.restoration import denoise_tv_chambolle
denoise_rocket = denoise_tv_chambolle(noisy_rocket, weight=0.1)
show_image(denoise_rocket, "Denoised rocket")
```

## Denoising using Bilateral Filter

```python
from skimage.restoration import denoise_bilateral
denoised_rocket_bil = denoise_bilateral(noisy_rocket)
show_image(denoised_rocket_bil, "Denoised rocket using Bilateral")
```
